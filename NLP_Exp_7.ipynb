7.1 Analysis of Sentiment and Subjectivity

```python
!pip install textblob
```

```python
from textblob import TextBlob
```

```python
text = input("Enter a sentence: ")

blob = TextBlob(text)

polarity = blob.sentiment.polarity
subjectivity = blob.sentiment.subjectivity

print("Polarity:", polarity)
print("Subjectivity:", subjectivity)

if polarity > 0:
    print("Sentiment: Positive")
elif polarity < 0:
    print("Sentiment: Negative")
else:
    print("Sentiment: Neutral")

if subjectivity > 0.5:
    print("The sentence is Subjective")
else:
    print("The sentence is Objective")
```

7.2 Sentiment Analysis on IMDB Movie Reviews

```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense
```

```python
vocab_size = 10000
max_length = 200

(x_train, y_train), (x_test, y_test) = imdb.load_data(
    num_words=vocab_size
)

print("Training samples:", len(x_train))
print("Testing samples:", len(x_test))
```

```python
x_train = pad_sequences(
    x_train,
    maxlen=max_length,
    padding='post',
    truncating='post'
)

x_test = pad_sequences(
    x_test,
    maxlen=max_length,
    padding='post',
    truncating='post'
)

print(x_train.shape)
print(x_test.shape)
```

```python
model = Sequential([
    Embedding(vocab_size, 16),
    GlobalAveragePooling1D(),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=512,
    validation_split=0.2
)
```

```python
loss, accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
prediction = model.predict(x_test[:1])

print("Prediction score:", prediction[0][0])

if prediction[0][0] >= 0.5:
    print("Predicted Sentiment: Positive")
else:
    print("Predicted Sentiment: Negative")

print("Actual Sentiment:",
      "Positive" if y_test[0] == 1 else "Negative")
```
